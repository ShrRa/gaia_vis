```python

df = pd.read_parquet(dp['dr3_sample_v1'])
```

```python
df[['l','b']].describe()
```

```python
df['parallax_distance'] = 1/df['parallax']
```

```python
# Try switching between different distances and None (which means unit distance), it's educational
#x, y, z = galactic_to_cartesian(df,r_colname='parallax_distance')
#x, y, z = galactic_to_cartesian(df,r_colname='distance_gspphot')
x, y, z = galactic_to_cartesian(df,r_colname=None)
mask = (abs(x)<10**3) & (abs(y)<10**3) & (abs(z)<10**3)

fig = go.Figure(data=[go.Scatter3d(
    x=x[mask], y=y[mask], z=z[mask],
    mode='markers',
    marker=dict(size=4, opacity=0.5,color=df['bp_rp'][mask], colorscale='plasma', colorbar=dict(title='bp_rp'))
)])
fig.update_layout(
    autosize=False,
    width=1000,
    height=800,
    margin=dict(
        l=50,
        r=50,
        b=20,
        t=20,
        pad=4)
    )
fig.show()
```
